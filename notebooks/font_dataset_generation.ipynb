```python
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont
```

```python
# Get the project directory
PROJECT_DIR = Path.cwd().parents[0]
```

```python
IMG_SIZE = 64
FONT_SIZE = 48
```

```python
def generate_centered_number_images(font_name: str, font_path: str, img_size: int, font_size: int) -> None:
    """
    Generates images with centered numbers from 1 to 9 and saves them to disk.

    Args:
        font_name: The name of the fond
        font_path: The path to the font on the system
        img_size: The size of the image
        font_size: The size of the font
    """

    font = ImageFont.truetype(font_path, font_size)

    output_dir = PROJECT_DIR / "data" / {font_name}
    output_dir.mkdir(parents=True, exist_ok=True)

    for i in range(1, 10):

        # Create the image (canvas)
        img = Image.new("RGB", (img_size, img_size), "white")
        draw = ImageDraw.Draw(img)

        text = str(i)

        # Find the width and height of the text bounding box
        left, top, right, bottom = draw.textbbox((0, 0), text, font=font)
        text_width = right - left
        text_height = bottom - top

        # Find the center position
        x = (img_size - text_width) / 2 - left
        y = (img_size - text_height) / 2 - top

        # Draw the text at the center of the image
        draw.text((x, y), text, font=font, fill="black")

        img.save(output_dir / f"{font_name}-sample-{i}.png")
```
